```python
import pandas as pd

hiring_full = pd.read_csv("../output/hiring_score/company_hiring_score.csv")  

companies_100k = pd.read_csv("./01_CompaniesSelected/UKcompanies_active_account_category_sample_100k.csv")
```

```python
# 统一格式：补齐到8位字符串
def normalise_cn(series):
    return (
        series.astype(str)
              .str.strip()
              .str.upper()                        # SC开头的公司大写
              .str.zfill(8)                       # 数字补前导零到8位
    )
```

```python

hiring_full["CompanyNumber"] = normalise_cn(hiring_full["CompanyNumber"])
companies_100k["CompanyNumber"] = normalise_cn(companies_100k["CompanyNumber"])

hiring_100k = companies_100k[["CompanyNumber"]].merge(
    hiring_full,
    on="CompanyNumber",
    how="left"
)
```

```python

print(f"100k has hiring score : {hiring_100k['company_hiring_score'].notna().sum():,}")
print(f"missing: {hiring_100k['company_hiring_score'].isna().sum():,}")
```

```python
hiring_100k.to_csv("../output/hiring_100k.csv", index=False, encoding="utf-8-sig")
```
