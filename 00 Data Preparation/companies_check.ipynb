```python
import pandas as pd

df = pd.read_csv('../output/UKcompanies_8_sectors_cleaned.csv')

df.shape
```

```python
missing_counts = df.isnull().sum()
print("missing：\n", missing_counts)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("\nmissing rate(%)：\n", missing_percentage.round(2)) # round(2) 保留两位小数
```

```python
missing_only = missing_counts[missing_counts > 0]
print(missing_only)
```
